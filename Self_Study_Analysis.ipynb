```python
import pandas as pd
import numpy as np

# DAILY SELF STUDY ANALYSIS PROJECT
data = {
    'Date': [
        '2026-09-02', '2026-09-05', '2026-09-09',
        '2026-09-13', '2026-09-17', '2026-09-21',
        '2026-09-25', '2026-09-29', '2026-10-03',
        '2026-10-06', '2026-10-10', '2026-10-14',
        '2026-10-18', '2026-10-22', '2026-10-27',
        '2026-11-01', '2026-11-05', '2026-11-09',
        '2026-11-14', '2026-11-18'
    ],
    'Subject': [
        'Python', 'Maths', 'Physics', 'English', 'Data Analysis',
        'Python', 'Maths', 'Python', 'Data Analysis', 'Physics',
        'Python', 'Maths', 'Python', 'Data Analysis', 'English',
        'Python', 'Maths', 'Data Analysis', 'Python', 'Physics'
    ],
    'Study_Place': [
        'Home', 'Library', 'Home', 'College', 'Library',
        'Home', 'Home', 'Library', 'Library', 'College',
        'Home', 'Home', 'Library', 'Library', 'College',
        'Home', 'Home', 'Library', 'Home', 'College'
    ],
    'Time_Slot': [
        'Evening', 'Morning', 'Night', 'Afternoon', 'Morning',
        'Evening', 'Night', 'Morning', 'Morning', 'Afternoon',
        'Evening', 'Night', 'Morning', 'Morning', 'Afternoon',
        'Evening', 'Night', 'Morning', 'Evening', 'Afternoon'
    ],
    'Study_Hours': [
        2.0, 1.5, 1.0, 1.0, 3.0,
        2.5, 1.5, 2.0, 3.5, 1.0,
        2.5, 1.0, 3.0, 3.0, 1.5,
        4.0, 1.5, 3.5, 5.0, 1.5
    ],
    'Break_Minutes': [
        15, 10, 10, 5, 20,
        15, 15, 10, 20, 10,
        15, 10, 20, 20, 10,
        25, 10, 25, 30, 10
    ],
    'Mobile_Hours': [
        3, 4, 5, 4, 1,
        3, 5, 2, 1, 4,
        3, 5, 1, 1, 4,
        2, 5, 1, 2, 4
    ],
    'Sleep_Hours': [
        7, 6, 5, 6, 8,
        7, 5, 7, 8, 6,
        7, 5, 8, 8, 6,
        7, 5, 8, 8, 6
    ],
    'Focus_Rating': [
        7, 6, 4, 5, 9,
        7, 4, 8, 9, 5,
        7, 4, 9, 9, 5,
        8, 4, 9, 8, 5
    ],
    'Topics_Completed': [
        4, 2, 1, 1, 6,
        4, 2, 4, 7, 1,
        5, 1, 6, 6, 2,
        8, 2, 7, 10, 2
    ],
    'Test_Score': [
        70, 62, 55, 58, 85,
        72, 56, 78, 88, 60,
        75, 54, 86, 84, 59,
        80, 57, 90, 87, 61
    ]
}

df = pd.DataFrame(data)

# Save the dataset as a CSV file (creates study_data.csv in Colab)
df.to_csv('study_data.csv', index=False)

# Download the CSV file to your computer (works in Google Colab)
try:
    from google.colab import files
    files.download('study_data.csv')
except ImportError:
    print("study_data.csv saved in the current folder")

# Import the CSV file back into a DataFrame
df = pd.read_csv('study_data.csv')

df['Date'] = pd.to_datetime(df['Date'])

# Productivity = topics completed per study hour (using NumPy)
df['Productivity'] = np.round(
    df['Topics_Completed'] / df['Study_Hours'], 2
)

df['Month'] = df['Date'].dt.month_name()

df['Weekday'] = df['Date'].dt.day_name()

weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
                 'Friday', 'Saturday', 'Sunday']


# QUESTION 1:
# What is the overall study time and how does it change over time?

print("\n1. OVERALL STUDY TIME")

print("Total Study Hours:",
      df['Study_Hours'].sum())

print("Average Study Hours per Session:",
      round(df['Study_Hours'].mean(), 2))

print("Maximum Study Hours in a Session:",
      df['Study_Hours'].max())

print("Total Topics Completed:",
      df['Topics_Completed'].sum())

print("\nMonthly Study Hours:")
print(df.groupby('Month', sort=False)['Study_Hours'].sum())


# QUESTION 2:
# Which subjects contribute the largest and smallest share of study time?

print("\n2. SUBJECT-WISE ANALYSIS")

subject_hours = df.groupby('Subject')['Study_Hours'].sum()

print(subject_hours.sort_values(ascending=False))

subject_share = (
    subject_hours / subject_hours.sum()
) * 100

print("\nSubject Study Time Share (%):")
print(np.round(subject_share, 2))


# QUESTION 3:
# Which study places and time slots perform best?

print("\n3. STUDY PLACE AND TIME SLOT ANALYSIS")

print(df.groupby('Study_Place').agg(
    Total_Sessions=('Subject', 'count'),
    Total_Hours=('Study_Hours', 'sum'),
    Average_Focus=('Focus_Rating', 'mean')
).sort_values(by='Average_Focus', ascending=False))

print("\nTime Slot Performance:")
print(df.groupby('Time_Slot').agg(
    Total_Sessions=('Subject', 'count'),
    Average_Focus=('Focus_Rating', 'mean'),
    Average_Productivity=('Productivity', 'mean')
).sort_values(by='Average_Focus', ascending=False))


# QUESTION 4:
# Which months and weekdays show stronger or weaker study activity?

print("\n4. MONTH AND WEEKDAY ANALYSIS")

print("\nMonthly Study Sessions:")
print(df.groupby('Month', sort=False)['Subject'].count())

print("\nWeekday Average Study Hours:")
weekday_hours = df.groupby('Weekday')['Study_Hours'].mean().reindex(weekday_order).dropna()
print(weekday_hours)

print("\nStrongest Weekday:", weekday_hours.idxmax())
print("Weakest Weekday:", weekday_hours.idxmin())


# QUESTION 5:
# How does average study time and focus vary across subjects and places?

print("\n5. AVERAGE STUDY TIME AND FOCUS")

print("\nSubject-wise Average Study Hours:")
print(df.groupby('Subject')['Study_Hours'].mean())

print("\nSubject-wise Average Focus:")
print(df.groupby('Subject')['Focus_Rating'].mean())

print("\nPlace-wise Average Study Hours:")
print(df.groupby('Study_Place')['Study_Hours'].mean())


# QUESTION 6:
# Are more mobile hours and less sleep associated with lower focus?

print("\n6. MOBILE USAGE AND SLEEP ANALYSIS")

print("Average Mobile Hours:",
      np.mean(df['Mobile_Hours']))

print("Average Sleep Hours:",
      np.mean(df['Sleep_Hours']))

correlation_mobile = np.corrcoef(
    df['Mobile_Hours'], df['Focus_Rating']
)[0, 1]

correlation_sleep = np.corrcoef(
    df['Sleep_Hours'], df['Focus_Rating']
)[0, 1]

print("Mobile Hours and Focus Correlation:",
      round(correlation_mobile, 2))

print("Sleep Hours and Focus Correlation:",
      round(correlation_sleep, 2))

correlation_score = np.corrcoef(
    df['Study_Hours'], df['Test_Score']
)[0, 1]

print("Study Hours and Test Score Correlation:",
      round(correlation_score, 2))


# QUESTION 7:
# Which subjects are studied frequently or for the longest time?

print("\n7. SUBJECT FREQUENCY AND HIGH-VALUE SESSIONS")

subject_analysis = df.groupby('Subject').agg(
    Total_Sessions=('Subject', 'count'),
    Total_Hours=('Study_Hours', 'sum'),
    Average_Test_Score=('Test_Score', 'mean')
)

print(subject_analysis.sort_values(
    by='Total_Hours', ascending=False
))

print("\nHigh-Value Sessions (Productivity >= 2 topics/hour):")
print(df[df['Productivity'] >= 2][
    ['Date', 'Subject', 'Study_Hours', 'Productivity']
])


# QUESTION 8:
# What actions could improve study efficiency without only studying longer?

print("\n8. STUDY IMPROVEMENT ANALYSIS")

print("Average Focus Rating:",
      round(np.mean(df['Focus_Rating']), 2))

print("Average Test Score:",
      round(np.mean(df['Test_Score']), 2))

print("\nLow Focus Sessions (Focus <= 5):")
print(df[df['Focus_Rating'] <= 5][
    ['Date', 'Subject', 'Time_Slot', 'Sleep_Hours', 'Mobile_Hours', 'Focus_Rating']
])

print("\nSessions with Less Sleep (< 6 hours):")
print(df[df['Sleep_Hours'] < 6][['Date', 'Subject', 'Sleep_Hours', 'Focus_Rating']])

print("\nSubject Needing More Attention:")
print(subject_hours.idxmin())

print("\nBest Time Slot for Studying:")
print(df.groupby('Time_Slot')['Focus_Rating'].mean().idxmax())

print("\nBest Place for Studying:")
print(df.groupby('Study_Place')['Focus_Rating'].mean().idxmax())

print("\nSUGGESTIONS:")
print("1. Difficult subjects ko best time slot mein padho.")
print("2. Sleep 7 ghante se kam na ho, aur mobile use kam rakho.")
print("3. Kam time wale subject ke liye weekly target banao.")


# FINAL DATA SUMMARY

print("\n========== FINAL STUDY DATA SUMMARY ==========")

print(df)

print("\nStatistical Summary:")
print(df.describe())

print("\nTotal Number of Records:", len(df))

print("Total Unique Subjects:",
      df['Subject'].nunique())

print("\nDAILY SELF STUDY ANALYSIS COMPLETED SUCCESSFULLY!")
```
